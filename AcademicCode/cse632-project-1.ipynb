```python
# importing the libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
# loading the dataset
df = pd.read_csv("Crime_Data_from_2020_to_Present.csv")
```

```python
# previewing the dataset
df.head()
```

```python
# previewing the column names and size
print(df.columns)
print(f"Size of the dataset is {df.shape}")
```

```python
# creating the heatmap of missing values
sns.heatmap(df.isna())
plt.title("Heatmap to show missing values")
plt.show()
```

```python
# creating table of missing values
missing_counts = df.isnull().sum()
missing_percentages = (df.isnull().sum() / len(df)) * 100
missing_value_df = pd.DataFrame({'Total': missing_counts, 'Percent': missing_percentages})
missing_value_df = missing_value_df.sort_values(by='Total', ascending=False)
missing_values_df = missing_value_df[missing_value_df['Total'] > 0]
missing_values_df
```

```python
# creating the graph of missing values
# plt.figure(figsize=(7,7))
plt.rcParams['figure.dpi'] = 300
plt.bar(missing_values_df.index, missing_values_df["Percent"])
plt.xticks(rotation=90)
plt.yticks(ticks=range(0,110,10), labels=[f"{x}%" for x in range(0,110,10)])
plt.ylabel("Percentage of missing values")
plt.ylabel("Column Names")
plt.title("Missing values in the dataset", fontweight="bold", fontsize="xx-large")
plt.show()
```
